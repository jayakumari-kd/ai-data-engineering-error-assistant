```python
import os

def list_files(path="."):
    return os.listdir(path)

tools = [
    {
        "name": "list_files",
        "description": "Lists files in a given directory",
        "input_schema": {
            "type": "object",
            "properties": {
                "path": {"type": "string"}
            },
            "required": ["path"]
        
        }
    }
]
```

```python
response=client.messages.create(model=model,
    max_tokens=100,
    tools=tools,
    messages=[
        {"role":"user",
        "content":"list all files in ."
}
])
```

```python
response
```

```python
tool_input = response.content[0].input
tool_output = list_files(tool_input["path"])
print(tool_output)
```

```python
# Step 1: Claude makes a tool call
response = client.messages.create(
    model=model,
    max_tokens=200,
    tools=tools,
    messages=[
        {"role":"user","content":"list all files in . and show me the output"}
    ]
)

# Step 2: You run your Python function
tool_input = response.content[0].input
tool_output = list_files(tool_input["path"])

# Step 3: Send the tool result back in a USER message
followup = client.messages.create(
    model=model,
    max_tokens=200,
    messages=[
        {"role": "assistant", "content": response.content},  # Claude’s tool call
        {
            "role": "user",
            "content": [
                {
                    "type": "tool_result",
                    "tool_use_id": response.content[0].id,  # link back to tool call
                    "content": str(tool_output),
                    "is_error": False
                }
            ]
        }
    ]
)

print(followup)
```
